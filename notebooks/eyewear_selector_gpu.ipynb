Bộ nhận diện kính RGB độc lập: MobileNetV3-Small. Không sửa model mắt gốc hay ngáp. CelebA gắn trực tiếp làm input, không xuất lại ảnh. YuNet và crop mặt theo runtime; preprocessing 128×128 được ghi trong manifest. Chọn epoch/ngưỡng bằng validation; test chỉ đo sau khi khóa. Chưa nghiệm thu trên camera USB với kính cận. Nguồn: https://mmlab.ie.cuhk.edu.hk/projects/CelebA.html


```python
import sys,subprocess
from pathlib import Path
from importlib.metadata import version
requirements=Path('/kaggle/working/requirements.txt');requirements.write_text('# Keep Kaggle\'s CUDA-enabled torch/torchvision; do not replace them with CPU wheels.\nultralytics==8.4.163\nonnx==1.17.0; python_version < "3.13"\nonnx==1.19.1; python_version >= "3.13"\nonnxruntime==1.20.1; python_version < "3.13"\nonnxruntime==1.23.2; python_version >= "3.13"\nnumpy==1.26.4; python_version < "3.13"\nnumpy==2.2.6; python_version >= "3.13"\nPillow==11.1.0\nopencv-python==4.11.0.86\n')
constraints=Path('/kaggle/working/gpu-constraints.txt');constraints.write_text(''.join(f'{n}=={version(n)}\n' for n in ('torch','torchvision')))
subprocess.check_call([sys.executable,'-m','pip','install','--quiet','--only-binary=:all:','-c',str(constraints),'-r',str(requirements)])
```

```python
script=next(Path('/kaggle/input').rglob('kaggle_eyewear_training.py'))
subprocess.check_call([sys.executable,'-u',str(script),'--root',str(script.parent),'--output','/kaggle/working/eyewear_model','--epochs','10'])
```
